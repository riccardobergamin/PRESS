## Scope

This vignette uses the possession-loss hazard $h$ and its difficulty bands, defined in the [composed hazard vignette](../06_composed_hazard_and_bands/06_composed_hazard_and_bands.ipynb), to describe how pressure evolves during an individual possession and to summarize how players and teams perform under pressure. It first defines three pressure patterns and illustrates each with an observed possession. It then introduces the expected-minus-observed score, which compares the failures of a player or team in elevated-risk situations with those predicted by $h$, and the report built on it. The results use ten A-League 2024/25 matches.


## Pressure patterns

PRESS describes three patterns of how pressure evolves during an individual possession. Two of them adapt concepts that SkillCorner defines in its pressure framework [@skillcorner2026], and the third is a project pattern.

* **Pressure escape.** SkillCorner's *Escaped Pressure* starts from a difficult pass reception and ends under lower overall pressure [@skillcorner2026escape]. The possession begins under pressure and ends with the pressure relieved.
* **Drawing pressure.** SkillCorner's *Drawing Pressure* starts from a low-pressure reception, lasts at least 2 s and ends under high pressure [@skillcorner2026drawing]. The possession begins comfortably and the pressure builds up while the player keeps the ball.
* **Sustained successful carrying.** The player carries the ball successfully for several seconds while the pressure stays elevated. It is close in spirit to SkillCorner's *Tight Space Retained Under Intense Pressure*, which also requires constrained space and difficult passing options [@skillcorner2026retained], conditions that we do not impose.

The patterns are retrospective indicators defined by this project. They do not reproduce SkillCorner's five-level degree, its separate difficulty flags or its proprietary categories, and they do not infer tactical intention.

### Identification with h

The unit is the **individual possession**, identified by `(match_id, possession_id)`. All indicators use the **14,919-state band-analysis cohort**, covering 7,265 verified outfield, non-shot possessions with a terminal canonical action, as described in the [composed-risk benchmark](../../benchmarks/06_composed_risk.qmd). This is distinct from the 15,085-state population used for the overall predictive metrics of h.

h is evaluated at the canonical checkpoints: one-second carry starts and pass releases. For a possession with several checkpoints, the initial state is its start. The final state is the release of its terminal pass or the start of its final failed carry. A loss at 2.4 s has checkpoints at 0, 1 and 2 s, with the final state at 2 s; a pass at 2.4 s has its final checkpoint at 2.4 s. The loss is an outcome of the final action, not an additional decision state.

**A micropossession shorter than one second has a single checkpoint, used as both its initial and final state by convention.** For a pass-only micropossession this is the release; for a failed carry it is the initial carry checkpoint. This conventional representative state need not coincide with the first tracking frame of a pass-only possession. Patterns requiring an initial state use possessions with that state under this convention.

* Elevated (Medium or High): $h\ge0.159104$.
* High: $h\ge0.292893$.
* Below Medium: $h<0.159104$.

The implementation uses the exact half-life cutoffs; the rounded values above give the same memberships. A completed pass is terminal individual nonloss; a failed pass or direct carry loss is terminal individual loss. Foul-suffered and clearance endings have no terminal canonical action and are outside this cohort; shots are excluded. Restart receptions may remain the start type of subsequent active-play possessions.

| Pattern | Operational definition | Eligible denominator |
|---|---|---|
| Pressure escape | Elevated initial h, final h below Medium, completed terminal pass, and duration at least 1 s. | All possessions with an elevated initial state. Losses and durations below 1 s count as non-escapes. |
| Drawing pressure | Reception with initial h below Medium, duration at least 2 s, and High at the final state. Terminal retention is not required. | All receptions with a low-risk initial state, including durations below 2 s as non-occurrences. Reception includes ordinary pass, free-kick, goal-kick, corner and throw-in reception start types. |
| Sustained successful carrying | At least two successful full one-second carries, each starting in an elevated state; they need not be consecutive and a later action may fail. | Possessions with at least one canonical carry attempted in an elevated state, including failed carries shorter than 1 s. |

Patterns need not be mutually exclusive. Intervals on their shares are **95% Wilson intervals over distinct possessions** [@wilson1927]. They are descriptive and do not account for dependence between possessions of the same match.

#### Case A: pressure escape in the defensive third

**Luis Felipe Gallegos Leiva, Western United–Auckland FC, 2025-05-03, 04:16.8–04:18.9.** Gallegos recovers an intercepted pass in his own defensive third with an opponent **1.59 m away**. He carries mainly laterally, gaining only 0.42 m along the attacking axis, and completes the outlet to goalkeeper **Alex Paulsen**. At release the nearest opponent is **3.88 m away**; h follows **42.24% → 6.95% → 7.81%**.

An associated engagement with Michael Ruhs corroborates the close pressure. The short lateral move and completed pass back to the goalkeeper illustrate an escape from immediate pressure during the build-up, rather than a forward break into open space [@skillcorner2026escape]. The last panel shows Paulsen receiving the pass. This is an observed pressure-relief sequence; body orientation and intention are not measured.

## Player and team report

The report uses frozen matching-fold held-out h for **14,919 canonical states, 7,265 individual possessions, 192 players and 12 teams** in ten A-League 2024/25 matches. Teams are represented in one to four matches. It describes this sample and is not a season-long ranking or a fitted model of player ability.

CIES reports pressure exposure and retention and discusses their dependence on playing context [@poli2024pressure]. A minimum-arrival-time study relates pressure at possession endpoints to progression and loss using a different pressure definition [@narizuka2026pressure]. These references motivate descriptive reporting rather than numerical comparisons.

h refers to the fragility of team possession. For player attribution, each canonical action is assigned to its current carrier; a completed pass ends that individual's involvement without ending the team sequence.

### Expected minus observed: E−O per 100 actions

For player $j$, let $A_j$ contain their canonical actions with $h_i\ge0.159104$, and let $y_i=1$ if the current action fails. The main score is

$$\mathrm{E{-}O}_j=100\,\frac{\sum_{i\in A_j}(h_i-y_i)}{|A_j|}=100\,(\bar h_j-\bar y_j).$$

A positive value means fewer failures than predicted; a negative value means more. The same definition applies to teams. No predictor is fitted for the report. Alongside the score, the tables report the number of elevated-risk actions, the number of distinct possessions containing them, represented matches, **mean h**, **observed failure frequency**, and the **share of High actions among elevated-risk actions**. If $A_j$ is empty, the score and these rates are undefined. A complete table retains those players with their counts.

### Uncertainty of the score

The intervals are **95% percentile intervals from 2,000 possession-bootstrap replicates**, with seed **20261030** [@efron1993bootstrap]. For each player, distinct `(match_id, possession_id)` blocks containing an action of $A_j$ are sampled with replacement, retaining all elevated-risk actions in a drawn block together. The score and the corresponding action means are recomputed with the resampled action denominator. Predictions and the elevated-risk membership remain fixed. Deterministic entity-specific random streams derived from the stated seed make the result independent of table ordering.

Players have only one to four represented matches, so match-level bootstrap is not the primary procedure. Teams use the same possession bootstrap; [match-bootstrap sensitivity](../../data/examples/pressure_report/team_match_bootstrap_sensitivity.csv) is retained separately. A one-match sensitivity interval is necessarily degenerate. Score intervals are shown only with **at least 20 distinct exposed possessions**. Below that count, the table shows the score and support without an interval.

The bootstrap preserves dependence within a possession. It does **not** capture dependence between possessions of the same match or team, nor model-estimation uncertainty. These intervals and scores remain exploratory. The optional O/E ratio is not used in this report.
