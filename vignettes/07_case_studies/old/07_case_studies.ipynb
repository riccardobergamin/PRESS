## Scope

This vignette uses the possession-loss hazard $h$ and its difficulty bands, defined in the [composed hazard vignette](../06_composed_hazard/06_composed_hazard.ipynb), to describe how pressure evolves during an individual possession and to summarize how players and teams perform under pressure. It first defines three pressure patterns and illustrates each with an observed possession. It then introduces the observed-minus-expected score, which compares the failures of a player or team in elevated-risk situations with those predicted by $h$, and the report built on it. The results use ten A-League 2024/25 matches.


## Pressure patterns

PRESS describes three patterns of how pressure evolves during an individual possession. Two of them adapt concepts that SkillCorner defines in its pressure framework [@skillcorner2026], and the third is a project pattern.

* **Pressure escape.** SkillCorner's *Escaped Pressure* starts from a difficult pass reception and ends under lower overall pressure [@skillcorner2026escape]. The possession begins under pressure and ends with the pressure relieved.
* **Drawing pressure.** SkillCorner's *Drawing Pressure* starts from a low-pressure reception, lasts at least 2 s and ends under high pressure [@skillcorner2026drawing]. The possession begins comfortably and the pressure builds up while the player keeps the ball.
* **Sustained successful carrying.** The player carries the ball successfully for several seconds while the pressure stays elevated. It is close in spirit to SkillCorner's *Tight Space Retained Under Intense Pressure*, which also requires constrained space and difficult passing options [@skillcorner2026retained], conditions that we do not impose.

The patterns are retrospective indicators defined by this project. They do not reproduce SkillCorner's five-level degree, its separate difficulty flags or its proprietary categories, and they do not infer tactical intention.

### Identification with h

Pressure is measured with h and its bands. *Elevated* means Medium or High ($h\ge0.159104$), and *High* means $h\ge0.292893$. The patterns are computed on possessions with at least one full second of carrying, so that the start state is a canonical checkpoint (one-second carries and the pass release). The start state is the first checkpoint of the possession and the end state its last. Since the start, the end and the outcome play different roles, each pattern has its own eligible denominator, and the patterns need not be mutually exclusive: a successful carrying sequence can also end in an escape.

| Pattern | Operational definition | Eligible denominator |
|---|---|---|
| Pressure escape | Elevated h at the start state, h below Medium at the end state, and no individual loss at the end. Recoveries count as well as receptions. | All represented possessions with an elevated start state; failed endings count as non-escapes. |
| Drawing pressure | Pass reception with h below Medium at the start state, duration of at least 2 s and High at the end state. Retention is reported separately and is not required for the pattern. | All represented low-risk pass receptions with a start state; a duration under 2 s is a non-occurrence. |
| Sustained successful carrying | At least two successful full one-second carries whose start states are elevated. The possession can still be lost. | Possessions with at least one attempted canonical carry in an elevated state. |

## Player and team report

The report summarizes **performance and exposure under pressure** on the ten represented A-League 2024/25 matches. It uses the frozen five-pair held-out predictions for 15,085 canonical states, 7,343 individual possessions, 192 players and 12 teams, with team coverage between 1 and 4 matches. With ten matches, the player values illustrate the format of the report and do not characterize individual players; they describe this sample and are not season-long rankings.

CIES reports both pressure exposure and retention and highlights their dependence on playing context [@poli2024pressure], which motivates the layout of the report. A recent study based on minimum arrival times compares pressure at the endpoints of a possession and relates it to progression and loss with a different pressure definition [@narizuka2026pressure].

The risk h refers to team possession. For player attribution, the report assigns each canonical action to the current carrier, and a completed pass ends that player's involvement without ending the team sequence.

### Observed minus expected

For player $j$, let $A_j$ be the canonical actions of $j$ with $h_i\ge0.159104$ (Medium or High), and let $y_i=1$ if action $i$ fails. The score is

$$\mathrm{OE}_j=100\,\frac{\sum_{i\in A_j}(h_i-y_i)}{|A_j|}=100\,(\bar h_j-\bar y_j),$$

where $\bar h_j$ is the mean predicted failure probability and $\bar y_j$ the observed failure rate over $A_j$. It counts the failures avoided per 100 elevated-risk actions relative to the prediction of h, and it is positive when fewer actions fail than predicted. The predictions are the held-out values of the matching outer fold, so no model is fitted for the report, and the same definition applies to teams. The number of actions $|A_j|$ measures the support of each value and is always shown beside the score.

### Uncertainty of the score

Intervals are 95% percentile intervals from 2,000 bootstrap replicates with a fixed seed [@efron1993bootstrap]. For a player, the possessions that contain actions of $A_j$ are resampled with replacement and all the actions of each drawn possession are kept, so the dependence among the actions of a possession is preserved. Resampling matches is not applicable, since each player appears in one to four matches. The predictions are fixed in every replicate, the set $A_j$ is determined by them, and $\mathrm{OE}_j$ is recomputed on the resampled actions. Teams use the same procedure on possessions, and resampling matches is used only as a sensitivity check. The intervals do not include the dependence among possessions of the same match or team, or the error of the model. With ten matches, the scores and their intervals are exploratory: they show how the report is built and do not support conclusions about individual players or teams.
