## Scope

This vignette applies the composed hazard $h$ and its fixed absolute bands to observed individual possessions and to ten A-League 2024/25 matches. The [composed hazard vignette](../06_composed_hazard/06_composed_hazard.ipynb) defines the score, the canonical checkpoints and the thresholds. The case studies are illustrative, and the population-level summaries state their own eligible denominators.


## Case studies

Three observed possessions show different pressure trajectories. Every canonical evaluated checkpoint has a pitch panel: the yellow star marks the carrier, the attacking direction is stated and velocity arrows are scaled drawing aids. Panel titles show h, its band, the weighted branch contributions, the distance of the nearest defender and the number of active annotated on-ball engagements (OBE). The values are held-out predictions of the matching outer-fold models at observed tracking states. Cases are selected for interpretation and not as a performance evaluation. Geometry, carrier movement, event origin, pass result, pass options and OBE timing were cross-checked against tracking and annotations, which do not observe body orientation or intention.


## Case A: escape after recovery {#pressure-escape-case}

**Nathaniel Atkinson, Melbourne City–Macarthur, 7 March 2025.** The possession starts
from a **pass interception**, not reception. An opponent engagement spans the
recovery and most of the possession. At the initial state, the nearest opponent is
2.69 m away; at the final pass state, the nearest is 6.38 m away and the annotated
engagement has ended. The carry-loss branch falls from 10.10% to 0.42%, and
expected pass failure from 42.40% to 5.66%. h decreases
**36.73%→13.52%→3.45%→3.03%**.

The endpoint is a completed 5.54 m lateral pass to Kai Trewin. Net longitudinal
progress is approximately −0.54 m: the interpretation is **relief from immediate
pressure and safer options**, not a progressive dribble. Spatial separation,
branch-risk reduction and the completed release jointly support the escape
label. The escape rule used here, defined in [Pressure patterns](#pressure-patterns), extends the SkillCorner concept [@skillcorner2026escape] to recoveries.

## Case B: drawing pressure {#drawing-pressure-case}

**Harry Van Der Saag, Brisbane–Perth, 21 December 2024,04:27.1–04:30.4.** He
receives with the nearest opponent 9.14 m away and no active annotated engagement.
He advances 7.27 m along the attacking axis. At release, three engagement events
are active and the nearest opponents are 4.24 m and 4.40 m away.

h rises **1.18%→3.72%→16.41%→34.98%**, while the carry and pass branches both
become riskier. The sequence lasts 3.3 s and ends with a completed 11.05 m lateral
pass to Adam Zimarino. These observations support an operational **drawing
pressure** pattern: a relatively free reception develops into engagement with
several opponents before release [@skillcorner2026drawing]. They do not identify
a conscious baiting intention. This is a successful action under increased
modeled risk.

## Case C: carrying during a close engagement {#carrying-case}

**Trent Ostler, Brisbane–Perth, 21 December 2024,45:13.0–45:19.5.** He receives
the ball with two opponents within 3 m. The nearest opponent stays between 1.14 m
and 1.90 m at all seven shown checkpoints; a single associated engagement is
active throughout. Six canonical one-second carries succeed before a completed
19.62 m lateral pass to Abdelelah Faisal. Net forward displacement is 9.50 m.

h is Medium/High during the six carries, then falls to Low at release:
39.91%,22.77%,17.51%,27.09%,24.43%,26.03%,10.54%. This is a good illustration of
**retaining control through sustained close engagement**. The last lower-risk
state does not mean the marker has disappeared. The number of annotated
passing options increases from 2 to 4, and expected pass failure decreases. Risk
relief and literal separation from a defender are distinct observations.

```python
import numpy as np
import pandas as pd
from pathlib import Path
from PRESS.utils.possession import hazard_band_thresholds, hazard_to_band

root = next(p for p in [Path.cwd(), *Path.cwd().parents] if (p / "_quarto.yml").is_file())
cases = pd.read_csv(root / "data/examples/hazard/cases.csv", dtype={"match_id": str}, keep_default_na=False)
recomputed = (1 - cases.p_pass) * cases.drive_loss + cases.p_pass * cases.pass_failure
assert np.allclose(recomputed, cases.h, atol=1e-12)
assert np.array_equal(hazard_to_band(cases.h), cases.band)
c = cases[cases.case.eq("C")].sort_values("frame_start")
medium_cut = hazard_band_thresholds()[4]
assert ((c.action.eq("ball_drive")) & (c.h >= medium_cut) & c.outcome.eq(1)).sum() == 6
print(cases[["case", "time_since_start", "action", "p_pass", "drive_loss",
             "pass_failure", "h", "band"]].to_string(index=False))
```

## Pressure patterns {#pressure-patterns}

Three report patterns describe how risk evolves during an individual possession, using the absolute bands of h. They are retrospective indicators defined by this project, adapted from football concepts of SkillCorner, and they do not reproduce its proprietary categories. The patterns need not be mutually exclusive: a successful carrying sequence can also end in an escape. Since the start, the end and the observed outcome play different roles, each statistic has its own eligible denominator.

SkillCorner's *Escaped Pressure* starts from a difficult pass reception and ends under lower overall pressure. *Drawing Pressure* starts from a low-pressure reception, lasts at least 2 s and ends under high pressure. *Tight Space Retained Under Intense Pressure* also requires constrained space and difficult passing options [@skillcorner2026escape; @skillcorner2026drawing; @skillcorner2026retained]. We adopt the concepts with our own h-based rules and do not reproduce SkillCorner's five-level degree or its separate difficulty flags.

h is computed at the canonical checkpoints, namely the one-second carries and the pass release. The *true start* is the state at the first instant of the individual possession and the *true end* the state at its last instant. A possession has no checkpoint at the true start when it ends in a pass before one full second of carrying, and none at the true end when it ends in a loss, since the final unsuccessful carry begins at the last whole second. The patterns that need these states use the possessions that have them. *Elevated* means Medium or High ($h\ge0.159104$).

| Pattern | Operational definition | Eligible denominator |
|---|---|---|
| Pressure escape | Elevated h at the true start, h below Medium at the true end, and no individual loss at the end. Recoveries count as well as receptions. | All represented possessions with an elevated true start; failed endings count as non-escapes. |
| Drawing pressure | Pass reception with h below Medium at the true start, duration of at least 2 s and High ($h\ge0.292893$) at the true end. Retention is reported separately and is not required for the pattern. | All represented low-risk pass receptions with a true start; a duration under 2 s is a non-occurrence. |
| Sustained successful carrying | At least two successful full one-second carries whose start states are elevated. The possession can still be lost. | Possessions with at least one attempted canonical carry in an elevated state. |

The patterns do not infer tactical intention and are not a classification model. Completion rates include the failed situations in the eligible denominator. The last successful stretch before a pass is not a full carry, and additional snapshots do not become extra actions or independent exposure units, as explained in the [composed hazard vignette](../06_composed_hazard/06_composed_hazard.ipynb).

## Player and team report

The report summarizes **performance and exposure under pressure** on the ten represented A-League 2024/25 matches. It uses the frozen five-pair held-out predictions for 15,085 canonical states, 7,343 individual possessions, 192 players and 12 teams, with team coverage between 1 and 4 matches. With ten matches, the player values illustrate the format of the report and do not characterize individual players; they describe this sample and are not season-long rankings.

CIES reports both pressure exposure and retention and highlights their dependence on playing context [@poli2024pressure], which motivates the layout of the report. A recent study based on minimum arrival times compares pressure at the endpoints of a possession and relates it to progression and loss with a different pressure definition [@narizuka2026pressure].

The risk h refers to team possession. For player attribution, the report assigns each canonical action to the current carrier, and a completed pass ends that player's involvement without ending the team sequence.

### Observed minus expected

For player $j$, let $A_j$ be the canonical actions of $j$ with $h_i\ge0.159104$ (Medium or High), and let $y_i=1$ if action $i$ fails. The score is

$$\mathrm{OE}_j=100\,\frac{\sum_{i\in A_j}(h_i-y_i)}{|A_j|}=100\,(\bar h_j-\bar y_j),$$

where $\bar h_j$ is the mean predicted failure probability and $\bar y_j$ the observed failure rate over $A_j$. It counts the failures avoided per 100 elevated-risk actions relative to the prediction of h, and it is positive when fewer actions fail than predicted. The predictions are the held-out values of the matching outer fold, so no model is fitted for the report, and the same definition applies to teams. The number of actions $|A_j|$ measures the support of each value and is always shown beside the score.


### Team exposure

Across the sample, 2,542 of the 7,343 individual possessions contain an elevated-risk state, and 1,632 of them end without individual loss (64.2%). ## Reproduce the report

The code below reads only public aggregate tables. No new model, threshold, calibration or player-specific parameter was fitted for the report.


```python
from pathlib import Path
import numpy as np
import pandas as pd
from scripts.build_pressure_report_figures import main as rebuild_pressure_figures
root = next(p for p in [Path.cwd(), *Path.cwd().parents] if (p / "_quarto.yml").is_file())
folder = root / "data/examples/pressure_report"
teams = pd.read_csv(folder / "team_summary.csv")
players = pd.read_csv(folder / "player_summary.csv")
assert teams.n_possessions.sum() == 7343
assert teams.n_pressure_possessions.sum() == 2542
assert teams.n_pressure_retained.sum() == 1632
assert np.allclose(teams.retention_pct, 100 * teams.n_pressure_retained / teams.n_pressure_possessions)
for pid in [14736, 51045]:
    row = players[players.provider_player_id.eq(pid)].iloc[0]
    assert np.isclose(row.retention_pct, 100 * row.n_pressure_retained / row.n_pressure_possessions)
    print(row[["player_name", "n_matches", "n_pressure_retained", "n_pressure_possessions", "retention_pct"]].to_dict())
print(teams[["team_name", "n_matches", "n_pressure_possessions", "retention_pct"]].to_string(index=False))
rebuild_pressure_figures()
```
