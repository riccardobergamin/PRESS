## Conceptual definition

These three report patterns describe **how risk evolves during individual
control**, using the absolute h bands. They are not required to be mutually
exclusive: successful carrying may also end in an escape. Initial state, end
state and observed outcome play different roles, so each statistic needs its
own eligible denominator.

SkillCorner's *Escaped Pressure* starts from a difficult pass reception and
ends with lower overall pressure. *Drawing Pressure* starts with a low-pressure
reception, lasts at least 2 s and ends under high pressure. Its *Tight Space
Retained Under Intense Pressure* also requires constrained space and difficult
passing options [@skillcorner2026escape; @skillcorner2026drawing;
@skillcorner2026retained]. We adopt the football concepts, with **our own
h-based operational rules**. We do not reproduce SkillCorner's proprietary
five-level degree or its separate difficulty flags.

## Definitions and denominators

Let $\tau=h_4=0.159104$ mark Medium/High and $\tau_H=h_2=0.292893$ mark High.

| Pattern | Operational definition | Eligible denominator |
|---|---|---|
| Pressure escape | True starting h≥τ, true ending h<τ and terminal individual nonloss. Recoveries as well as receptions are allowed. | All represented possessions with a true observed starting h≥τ; failed endings count as non-escapes. |
| Drawing pressure | Pass reception with true starting h<τ, duration≥2 s and true ending h≥τH. Retention is reported separately; it is not required to detect the pattern. | All represented low-risk pass receptions with a true observed start; duration<2 s is a non-occurrence. |
| Sustained successful carrying under pressure | At least two successful full one-second carries whose start states have h≥τ. The later possession can still be lost. | Report against possessions with any attempted canonical carry under elevated h; do not mistake this selected pattern for completion rate of all attempts. |

These are retrospective descriptive patterns. They do not infer tactical
intention or form a classification model. For completion rates, include failed
situations in the eligible denominator. Defining a category by successful escape
and then reporting its retention as performance would be circular.

The last successful microtail before a pass is not a full carry. Additional
snapshots can enrich a profile but do not become extra training actions or
independent exposure units. The [hazard vignette](../05_composed_hazard_and_bands/05_composed_hazard_and_bands.ipynb)
explains the canonical convention.

## Endpoint coverage and missing classifications

The frozen benchmark represents 7,343 individual controls. True starting h is
available for 6,097 and true ending h for 6,847; 5,601 have both. Missing endpoints
are **not replaced by the first/last available canonical row**. A loss rules out
a successful escape even if its final h is unavailable; otherwise unresolved
endpoint-dependent patterns remain unknown. Counts and logical bounds are
reported rather than silently treating missing observations as negatives.

* Escape: 239 observed cases among 1,562 eligible starts, 22 unresolved; the full
  eligible-denominator frequency lies between 15.30% and 16.71%.
* Drawing: 38 observed cases among 3,454 eligible receptions, 95 unresolved;
  frequency bounds 1.10%–3.85%.
* Sustained successful carrying: 269 possessions contain at least two such carries,
  among 1,113 possessions with an elevated-risk attempted carry.

These frequencies describe the represented canonical sample and these rules,
not the full league or a comparison with SkillCorner's reported rates.

## Case C: carrying during a close engagement {#carrying-case}

**Trent Ostler, Brisbane–Perth, 21 December 2024,45:13.0–45:19.5.** He receives
the ball with two opponents within 3 m. The nearest opponent stays between 1.14 m
and 1.90 m at all seven shown checkpoints; a single associated engagement is
active throughout. Six canonical one-second carries succeed before a completed
19.62 m lateral pass to Abdelelah Faisal. Net forward displacement is 9.50 m.

h is Medium/High during the six carries, then falls to Low at release:
39.91%,22.77%,17.51%,27.09%,24.43%,26.03%,10.54%. This is a good illustration of
**retaining control through sustained close engagement**. The last lower-risk
state does not mean the marker has disappeared. The number of annotated
passing options increases from 2 to 4, and expected pass failure decreases. Risk
relief and literal separation from a defender are distinct observations.
