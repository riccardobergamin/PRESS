## Conceptual definition and sample

The report closes the portfolio with **exposure and performance under pressure**
on the ten represented A-League 2024/25 matches. It uses frozen five-pair OOF
predictions for 15,085 canonical states, 7,343 individual possessions, 192 players
and 12 teams. Team coverage ranges from 1 to 4 matches. These are sample profiles,
not full-season rankings.

CIES reports both pressure exposure and retention, highlighting their dependence
on playing context [@poli2024pressure]. That motivates the report layout; its
large multi-league study and proprietary pressure metric are not a numerical
benchmark for our sample. Temporal endpoint comparisons and links between
pressure, progression and loss are also explored in a recent minimum-arrival-time
study [@narizuka2026pressure], using a different pressure definition.

The risk h refers to team possession. For player attribution, the report groups
its observed values into the current carrier's individual controls. Passing
successfully completes that player's situation without terminating the team
sequence.

## Primary score: completion relative to situations faced

A pressure situation is **one individual possession with at least one observed
canonical h≥0.159104**, i.e. Medium/High. Count it once regardless of how many
checkpoints it contains. Let $N_j$ be the number of such situations for player j
and $K_j$ the number ending without individual loss:

$$\boxed{\mathrm{Retention}_j=100\,K_j/N_j.}$$

This percentage normalizes for opportunity volume. It does not normalize away
all differences in situation difficulty or tactical assignment; show exposure,
band profiles and sample size beside it. The same definition applies to teams.
Do not define the denominator using only successful escapes or only successful
carries. These percentages describe terminal outcomes, unlike h's action-level
calibration target.

The retained label follows the agreed framework: completed pass or foul suffered
is nonloss; failed pass, direct possession loss or clearance is individual loss.
Clearance remains an operational convention and can be tactically appropriate.
Shot-ending possessions and unverifiable terminal outcomes are outside this
cohort. Thus the percentage is not a universal measure of footballing quality.

The exposure is **observed canonical-state exposure**, not continuous time under
pressure: a short unsampled high-risk interval can be missed. Overall retention
uses all 7,343 represented parents, not just the 5,601 with both true endpoint h
values. Endpoint-dependent patterns are reported separately with unknowns.

## Team exposure and retention

Across the sample, 2,542 individual possessions contain an elevated-risk state;
1,632 end without individual loss: **64.2% retention**. The left panel shows each
possession's highest observed band, so long controls do not receive multiple
counts. The right panel shows retention within Medium/High-exposed possessions,
with the numerator/denominator and number of represented matches visible.

![Team exposure and terminal retention in the ten-match sample.](../../figures/vignette_09_league_report.png)

Wellington has the largest observed exposure share in this sample (44.6%).
Melbourne City retains 72.4% of its pressure-exposed controls (184/254), compared
with 52.3% for Western United (124/237). Both have two represented matches.
These are descriptive contrasts; opponent, field position, choice and the
severity of exposure still differ. They are not a season-wide ordering of teams.

The intervals are **95% Wilson binomial reference intervals** over distinct
individual possessions [@wilson1927]. Shared team/match dependence and model
uncertainty are not included; they must not be read as uncertainty in stable
season-long skill. The fixed-band validation uses a separate whole-match
bootstrap [@efron1993bootstrap].

## Exploratory comparison with published reports

The CIES league report contextualizes retention by pressure exposure and playing
position across a much larger multi-league sample [@poli2024pressure]. It
provides a reference for what to report, rather than matched numerical rates:
its players, pressure metric, qualifying opportunities and coverage differ.
Our completion percentage remains an explicitly defined ten-match statistic.

A same-season football reference is the **2024/25 PFA A-League Men Team of the
Season**: Louis Verstraete is in the starting midfield and Anthony Cáceres in
the substitutes [@pfa2025team]. Atkinson, used in the escape example, is also
in the starting XI. These are peer-voted, broad performance recognitions;
they do not measure pressure retention or validate our ranking. The comparison
is deliberately exploratory: the selected profiles are compatible with wider
recognition, but ten matches cannot establish stable pressure-handling skill.
Auckland's account [@auckland2025pfa] and Sydney's on-ball description
[@sydney2024international] provide additional qualitative player context.

### Optional action-level observed–expected companion

For canonical actions evaluated under elevated h, a different summary is
$100\sum_i(h_i-y_i)/N_{actions}$, with $y_i$ the action-level failure label.
It is positive when fewer actions fail than predicted. Verstraete's value is
+8.10 and Cáceres' +10.28 per 100 actions in the frozen sample. These use 74 and 72 actions,
respectively; they are **not** the 38/52 and 32/44 possession-retention percentages.
The High-band pessimism and other calibration error affect this companion.
It is secondary, not a substitute for the agreed primary completion percentage.

## Reproduce the report

The code below reads only public aggregate tables. Preparing them used stable
provider player IDs, not match-local shirt numbers, and joined every outcome
once per match/individual possession. No new model, threshold, calibration,
player-specific parameter or production-weight change was fitted for this report.


```
from pathlib import Path
import numpy as np
import pandas as pd
from scripts.build_pressure_report_figures import main as rebuild_pressure_figures
root = next(p for p in [Path.cwd(), *Path.cwd().parents] if (p / "_quarto.yml").is_file())
folder = root / "data/examples/pressure_report"
teams = pd.read_csv(folder / "team_summary.csv")
players = pd.read_csv(folder / "player_summary.csv")
assert teams.n_possessions.sum() == 7343
assert teams.n_pressure_possessions.sum() == 2542
assert teams.n_pressure_retained.sum() == 1632
assert np.allclose(teams.retention_pct, 100 * teams.n_pressure_retained / teams.n_pressure_possessions)
for pid in [14736, 51045]:
    row = players[players.provider_player_id.eq(pid)].iloc[0]
    assert np.isclose(row.retention_pct, 100 * row.n_pressure_retained / row.n_pressure_possessions)
    print(row[["player_name", "n_matches", "n_pressure_retained", "n_pressure_possessions", "retention_pct"]].to_dict())
print(teams[["team_name", "n_matches", "n_pressure_possessions", "retention_pct"]].to_string(index=False))
rebuild_pressure_figures()
```
