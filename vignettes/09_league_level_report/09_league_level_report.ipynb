# League-level report

The reporting unit is the league sample: category shares and one or two possession-level case studies per category. The figure is generated from a labeled synthetic cohort and demonstrates the report layout. **design choice**

## Model and inputs

For a real-data report, first compute one mutually exclusive category per team possession. The league share is category count divided by all eligible team possessions. Case studies are selected by reproducible rules from that category and display its band trajectory, cumulative survival curve, and event context, including SkillCorner engagement where present. No team- or player-level rankings are part of this report. **design choice**.

## Usage and visual

The following code uses deterministic synthetic inputs. Any printed quantities and figure values are computed by this cell; no match-level data are embedded.

## Component scope and model limits

Aggregation scope and case-selection policy are **design choice**.

The included report figure is synthetic because no real-match data are redistributed. A league-level report requires local data and a completed team-possession table; the report is not a team- or player-level ranking. **scope**


```python
import numpy as np
import matplotlib.pyplot as plt
from pathlib import Path
categories=["Instant High-Risk Loss","Pressing to Loss","De-escalated but Lost","Escaped Risk","Retained Under Intense Risk","Attracted risk","Unexpected Loss","Ordinary/Low-Risk"]
# Fixed-seed synthetic category labels; all values are generated here.
rng=np.random.default_rng(2026); counts=rng.multinomial(120,np.array([.08,.14,.11,.12,.10,.08,.12,.25])); shares=counts/counts.sum()
fig,ax=plt.subplots(figsize=(9,4)); ax.barh(categories,shares,color="#4c78a8"); ax.set(xlabel="League-level share (synthetic cohort)",title="Typology report layout · synthetic data"); fig.tight_layout()
out=Path("figures/vignette_09_league_report.png"); out.parent.mkdir(exist_ok=True); fig.savefig(out,dpi=150); fig.savefig(out.with_suffix(".svg"))
print({name:float(share) for name,share in zip(categories,shares)})
```
